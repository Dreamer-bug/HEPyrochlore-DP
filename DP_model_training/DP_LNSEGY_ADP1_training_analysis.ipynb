```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

plt.figure(figsize=[8,6])

# 读取文件并处理数据
with open("lcurve.out") as f:
    headers = f.readline().split()[1:]

lcurve = pd.DataFrame(
    np.loadtxt("lcurve.out"), columns=headers
)

# 定义要绘制的曲线
legends = ["rmse_e_val", "rmse_e_trn", "rmse_f_val", "rmse_f_trn"]

# 对每条曲线应用滚动窗口平滑
window_size = 50  # 滚动窗口大小，可以根据需要调整

for legend in legends:
    smoothed_data = lcurve[legend].rolling(window=window_size).mean()  # 使用滚动窗口的均值进行平滑
    plt.loglog(lcurve["step"], smoothed_data, label=legend)

# 设置图例、轴标签等
plt.legend()
plt.xlim([0, 500000])
plt.ylim([10e-6, 5])
plt.xlabel("Training steps")
plt.ylabel("Loss")
plt.show()
```

```python
import pandas as pd
# 文件路径
file_path = 'test_results/Yb.f.out'
# 读取文件并解析为数据框
data = pd.read_csv(file_path, delim_whitespace=True, comment='#', names=['data_fx', 'data_fy', 'data_fz', 'pred_fx', 'pred_fy', 'pred_fz'])
# 将 data_e 列读取为列表
data_fx_list = data['data_fx'].tolist()
data_fy_list = data['data_fy'].tolist()
data_fz_list = data['data_fz'].tolist()
# 将 pred_e 列读取为列表
pred_fx_list = data['pred_fx'].tolist()
pred_fy_list = data['pred_fy'].tolist()
pred_fz_list = data['pred_fz'].tolist()
# 输出两个列表的前五个元素
print("data_f_list:", data_fx_list)
print("pred_f_list:", pred_fx_list)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as font_manager

# 创建图形和坐标轴对象
fig, ax = plt.subplots(1, 1, figsize=(8, 6), frameon=False, tight_layout=True)

# 假设data_fx_list, data_fy_list, data_fz_list, pred_fx_list, pred_fy_list, pred_fz_list已经定义
# 合并数据
data_f = data_fx_list + data_fy_list + data_fz_list
pred_f = pred_fx_list + pred_fy_list + pred_fz_list

# 绘制散点图
ax.scatter(data_f, pred_f)

# 绘制对角线
x_range = np.linspace(plt.xlim()[0], plt.xlim()[1])
ax.plot(x_range, x_range, "r--", linewidth=1.5)

# 设置坐标轴范围
plt.xlim(-12, 12)
plt.ylim(-12, 12)

# 设置坐标轴标签
ax.set_ylabel("Predicted Force [eV/Å]", fontsize=36, 
              fontproperties=font_manager.FontProperties(fname='times.ttf'))
ax.set_xlabel("Force of DFT [eV/Å]", fontsize=36, 
              fontproperties=font_manager.FontProperties(fname='times.ttf'))

# 设置x轴刻度和字体

ax.set_xticks([-12, -8, -4, 0, 4, 8, 12])
ax.set_xticklabels([-12, -8, -4, 0, 4, 8, 12],fontproperties=font_manager.FontProperties(fname='times.ttf'),fontsize=26)
ax.set_yticks([-12, -8, -4, 0, 4, 8, 12])
ax.set_yticklabels([-12, -8, -4, 0, 4, 8, 12],fontproperties=font_manager.FontProperties(fname='times.ttf'),fontsize=26)

rmse = np.sqrt(np.mean((np.array(pred_f) - np.array(data_f))**2))
print("RMSE:", rmse)
# 添加 标签
ax.text(0, -10, f"RMSE={rmse:.3e}", fontsize=28, fontproperties=font_manager.FontProperties(fname='times.ttf'))

# 显示图形
plt.show()

# 保存图像，取消注释以下行以保存
fig.savefig(f'test_figures/Yb_f.png', transparent=True, dpi=300)
```
